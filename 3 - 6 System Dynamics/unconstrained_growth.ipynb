```python
simulationLength = 20
population = 100
growthRate = 0.1
changeInTime = 0.1
numIterations = int(simulationLength / changeInTime)

for iteration in range(numIterations):
    growth = growthRate * population
    growthOverTime = growth * changeInTime
    population = population + growthOverTime
    print(f"Time = {(iteration + 1) * changeInTime:.1f}\t :\t Population = {population:.2f}")
```
